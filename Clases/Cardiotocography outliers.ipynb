### Z-Score

```python
#distancias
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import zscore

# 3. Calcular las distancias euclídeas de cada registro al vector promedio
# np.linalg.norm con axis=1 calcula la magnitud del vector diferencia para cada fila
distancias_euclideas = np.linalg.norm(X_Escalado - vector_promedio, axis=1)

# 4. Convertir las distancias a Z-scores
# Esto mide a cuántas desviaciones estándar del promedio se aleja la distancia de cada punto
z_scores = zscore(distancias_euclideas)
```

```python
print(f"Curtosis de las distancias: {pd.DataFrame(z_scores).kurtosis()}")
# Mostrar un resumen de los outliers encontrados con Z > 3
outliers = z_scores[z_scores > 3]
print(f"\nSe encontraron {len(outliers)} registros con un Z-score mayor a 3.")
```

```python
#Tabla de confusion
from sklearn.metrics import confusion_matrix
y_true=(y.NSP>1).astype(int)
y_pred = (z_scores > 3).astype(int)
cm = confusion_matrix(y_true, y_pred)
tn, fp, fn, tp = cm.ravel()

print("--- Resumen de la Clasificación ---")
print(f"Verdaderos Positivos (TP): {tp} (Anomalías reales detectadas)")
print(f"Verdaderos Negativos (TN): {tn} (Registros normales clasificados correctamente)")
print(f"Falsos Positivos (FP): {fp} (Falsas alarmas: normales etiquetados como outliers)")
print(f"Falsos Negativos (FN): {fn} (Anomalías que el modelo NO detectó)")

# Métricas clave en detección de anomalías
recall = tp / (tp + fn) if (tp + fn) > 0 else 0
precision = tp / (tp + fp) if (tp + fp) > 0 else 0

print(f"\nRecall (Sensibilidad): {recall:.2%} (De todos los outliers reales, ¿cuántos atrapamos?)")
print(f"Precision: {precision:.2%} (De todos los que predijimos como outliers, ¿cuántos lo eran realmente?)")
```

```python
plt.figure(figsize=(10, 6))
plt.hist(z_scores, bins="fd", color='skyblue', edgecolor='black', alpha=0.7)
plt.title('Distribución de Z-scores (Distancia Euclídea al Centro)', fontsize=14)
plt.xlabel('Z-score', fontsize=12)
plt.ylabel('Frecuencia (Cantidad de Registros)', fontsize=12)
plt.legend()
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
# Añadir líneas de referencia para umbrales comunes de detección de outliers (Z = 2 y Z = 3)
plt.axvline(x=2, color='orange', linestyle='--', linewidth=2, label='Z = 2 (Alerta)')
plt.axvline(x=3, color='red', linestyle='--', linewidth=2, label='Z = 3 (Outlier Fuerte)')

plt.show()
```

> ¿Qué tan validos son los límites de 2 o 3 desv estándar? ¿en qué supuestos se basan?

```python
from scipy.stats import anderson
# 1. Ejecutar la prueba de Anderson-Darling
# El parámetro dist='norm' indica que estamos probando contra una distribución normal.
resultado_ad = anderson(z_scores, dist='norm')

print("--- Prueba de Normalidad Anderson-Darling ---")
print(f"Estadístico de Prueba (A2): {resultado_ad.statistic:.4f}\n")

print("Resultados por Nivel de Significancia (Alfa):")
print("-" * 55)
print(f"{'Significancia (%)':<20} | {'Valor Crítico':<15} | {'¿Es Normal?'}")
print("-" * 55)
# 2. Iterar sobre los valores críticos para interpretar el resultado
# La prueba de A-D retorna múltiples valores críticos para diferentes niveles de confianza (15%, 10%, 5%, 2.5%, 1%)
es_normal = True

for i in range(len(resultado_ad.critical_values)):
    nivel_sig = resultado_ad.significance_level[i]
    valor_critico = resultado_ad.critical_values[i]
    
    # Si el estadístico calculado es mayor al valor crítico, rechazamos la hipótesis nula (no es normal)
    if resultado_ad.statistic > valor_critico:
        conclusion = "NO (Se rechaza H0)"
        es_normal = False
    else:
        conclusion = "SÍ (No se rechaza H0)"
        
    print(f"{nivel_sig:<20} | {valor_critico:<15.3f} | {conclusion}")

print("-" * 55)
```

### Tukey Fences

```python
# Calcular Cuartiles e IQR (Rango Intercuartílico)
q1 = np.percentile(distancias_euclideas, 25)
q3 = np.percentile(distancias_euclideas, 75)
iqr = q3 - q1
# El multiplicador clásico de Tukey es 1.5. (Para outliers "extremos", a veces se usa 2.0)
limite_inferior = q1 - 2 * iqr
limite_superior = q3 + 2 * iqr

# 3. Identificar los outliers
outliers_inferiores = distancias_euclideas[distancias_euclideas < limite_inferior]
outliers_superiores = distancias_euclideas[distancias_euclideas > limite_superior]

print("--- Análisis de Tukey Fences (IQR) ---")
print(f"Q1 (25%): {q1:.2f}")
print(f"Q3 (75%): {q3:.2f}")
print(f"IQR: {iqr:.2f}")
print(f"Límite Inferior: {limite_inferior:.2f}")
print(f"Límite Superior: {limite_superior:.2f}")
print(f"Outliers encontrados: {len(outliers_superiores)} superiores, {len(outliers_inferiores)} inferiores.")

# 4. Crear el Boxplot
plt.figure(figsize=(10, 4))

# plt.boxplot ya usa el multiplicador 1.5*IQR por defecto (parámetro whis=1.5)
plt.boxplot(distancias_euclideas, vert=False, patch_artist=True,
            boxprops=dict(facecolor='lightblue', color='blue'),
            medianprops=dict(color='red', linewidth=2),
            flierprops=dict(marker='o', markerfacecolor='orange', markersize=6, alpha=0.7))

# Dibujar explícitamente las vallas (Fences) de Tukey para mayor claridad visual
plt.axvline(limite_inferior, color='green', linestyle='--', linewidth=2, label=f'Fence Inferior ({limite_inferior:.1f})')
plt.axvline(limite_superior, color='red', linestyle='--', linewidth=2, label=f'Fence Superior ({limite_superior:.1f})')

plt.title('Boxplot de Distancias Euclídeas (Tukey Fences)', fontsize=14)
plt.xlabel('Distancia', fontsize=12)
plt.yticks([1], ['Distancias'])
plt.legend()
plt.tight_layout()

plt.show()
```

```python
mascara_outliers = (distancias_euclideas < limite_inferior) | (distancias_euclideas > limite_superior)
y_pred_tukey = mascara_outliers.astype(int)
cm = confusion_matrix(y_true, y_pred_tukey)
tn, fp, fn, tp = cm.ravel()

print("--- Resumen de la Clasificación ---")
print(f"Verdaderos Positivos (TP): {tp} (Anomalías reales detectadas)")
print(f"Verdaderos Negativos (TN): {tn} (Registros normales clasificados correctamente)")
print(f"Falsos Positivos (FP): {fp} (Falsas alarmas: normales etiquetados como outliers)")
print(f"Falsos Negativos (FN): {fn} (Anomalías que el modelo NO detectó)")

# Métricas clave en detección de anomalías
recall = tp / (tp + fn) if (tp + fn) > 0 else 0
precision = tp / (tp + fp) if (tp + fp) > 0 else 0

print(f"\nRecall (Sensibilidad): {recall:.2%} (De todos los outliers reales, ¿cuántos atrapamos?)")
print(f"Precision: {precision:.2%} (De todos los que predijimos como outliers, ¿cuántos lo eran realmente?)")
```

# Modelos basados en agrupamiento

## k-means

https://scikit-learn.org/stable/modules/clustering.html

En vez de tomar la distancia al centro de toda la masa de datos, se agrupan los datos y se mide la distancia de cada punto al grupo que supuestamente pertenece.

```python
from sklearn.cluster import KMeans
#Configurar y entrenar K-Means
k = 10 #Grupos
kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
kmeans.fit(X_Escalado)

X_mat = np.asarray(X_Escalado)
n_samples, n_features = X_mat.shape
n_clusters = kmeans.n_clusters
# 1. Global centroid (grand mean)
global_centroid = np.mean(X_mat, axis=0)

# 2. Within-Cluster Sum of Squares (SSW / Inertia)
ss_within = kmeans.inertia_

# 3. Between-Cluster Sum of Squares (SSB)
cluster_centers = kmeans.cluster_centers_
labels = kmeans.labels_

ss_between = 0.0
for k in range(n_clusters):
    n_k = np.sum(labels == k)  # Size of cluster k
    centroid_k = cluster_centers[k]
    ss_between += n_k * np.sum((centroid_k - global_centroid) ** 2)

# 4. Total Sum of Squares (SST)
ss_total = ss_within + ss_between

# 5. Variances (Mean Squares using Degrees of Freedom)
df_within = n_samples - n_clusters
df_between = n_clusters - 1
df_total = n_samples - 1

var_within = ss_within / df_within
var_between = ss_between / df_between
var_total = ss_total / df_total

# 6. Percentage of Variance Explained (Eta-squared)
pct_var_explained = (ss_between / ss_total) * 100

# Print Results
print("--- K-Means Variance Decomposition ---")
print(f"Total Sum of Squares (SST):   {ss_total:12.4f}  (df = {df_total})")
print(f"Within Sum of Squares (SSW):  {ss_within:12.4f}  (df = {df_within})")
print(f"Between Sum of Squares (SSB): {ss_between:12.4f}  (df = {df_between})\n")

print(f"Within-Group Variance (MSW):  {var_within:12.4f}")
print(f"Between-Group Variance (MSB): {var_between:12.4f}")
print(f"Total Variance (MST):         {var_total:12.4f}\n")

print(f"Variance Explained:           {pct_var_explained:12.2f}%")
```

```python
#Calcular la distancia de cada punto a su centroide
distancias = []
X_np = np.array(X_Escalado)
for i in range(len(X_Escalado)):
    centroide = kmeans.cluster_centers_[kmeans.labels_[i]]
    # Distancia Euclidiana
    dist = np.linalg.norm(X_np[i] - centroide)
    distancias.append(dist)

#Definir el umbral de anomalía
# Por ejemplo, consideramos anómalos al 1% de los puntos con mayor distancia
umbral = np.percentile(distancias, 99)
y_pred_kmeans = distancias > umbral

n_anomalias = y_pred_kmeans.sum()

print("--- Resultados de K-Means para Anomalías ---")
print(f"Total de registros: {len(X)}")
print(f"Umbral de distancia (Percentil 99): {umbral:.2f}")
print(f"Anomalías detectadas: {n_anomalias} ({(n_anomalias/len(X))*100:.2f}%)")
```

```python
cm = confusion_matrix(y_true, y_pred_kmeans)
tn, fp, fn, tp = cm.ravel()

print("--- Resumen de la Clasificación ---")
print(f"Verdaderos Positivos (TP): {tp} (Anomalías reales detectadas)")
print(f"Verdaderos Negativos (TN): {tn} (Registros normales clasificados correctamente)")
print(f"Falsos Positivos (FP): {fp} (Falsas alarmas: normales etiquetados como outliers)")
print(f"Falsos Negativos (FN): {fn} (Anomalías que el modelo NO detectó)")

# Métricas clave en detección de anomalías
recall = tp / (tp + fn) if (tp + fn) > 0 else 0
precision = tp / (tp + fp) if (tp + fp) > 0 else 0

print(f"\nRecall (Sensibilidad): {recall:.2%} (De todos los outliers reales, ¿cuántos atrapamos?)")
print(f"Precision: {precision:.2%} (De todos los que predijimos como outliers, ¿cuántos lo eran realmente?)")
```

## Modelos basados en proximidad 
Los algoritmos basados en proximidad (o distancia a vecinos) evalúan el grado de aislamiento de un punto midiendo la distancia a sus k vecinos más cercanos. A diferencia del enfoque global o por grupos, la detección basada en proximidad no asume un centroide: un punto es una anomalía si sus vecinos más cercanos están demasiado lejos.

### k nearest neighbours (k-nn)

```python
from sklearn.neighbors import NearestNeighbors
# Configurar el modelo k-NN
# Solicitamos k+1 vecinos porque el primer vecino retornado es el propio punto (distancia = 0)
k = 10
nn = NearestNeighbors(n_neighbors=k + 1, algorithm='auto')
nn.fit(X_Escalado)

# Calcular distancias a los k vecinos
distancias, indices = nn.kneighbors(X_Escalado)

# Excluimos la primera columna (distancia a sí mismo)
distancias_vecinos = distancias[:, 1:]

# Score de Proximidad: Distancia media a los k vecinos
score_proximidad = np.mean(distancias_vecinos, axis=1)

# Establecer un umbral de anomalía
# Por ejemplo, etiquetamos el top 1% con mayor score de proximidad (mayor aislamiento)
umbral = np.percentile(score_proximidad, 99)
y_pred_knn = score_proximidad > umbral

n_anomalias = y_pred_knn.sum()

print("--- Resultados del Algoritmo Basado en Proximidad (k-NN) ---")
print(f"Total de registros: {len(X_Escalado)}")
print(f"Número de vecinos (k): {k}")
print(f"Umbral de distancia (Percentil 95): {umbral:.4f}")
print(f"Anomalías detectadas: {n_anomalias} ({(n_anomalias/len(X))*100:.2f}%)")
```

```python
cm = confusion_matrix(y_true, y_pred_knn)
tn, fp, fn, tp = cm.ravel()

print("--- Resumen de la Clasificación ---")
print(f"Verdaderos Positivos (TP): {tp} (Anomalías reales detectadas)")
print(f"Verdaderos Negativos (TN): {tn} (Registros normales clasificados correctamente)")
print(f"Falsos Positivos (FP): {fp} (Falsas alarmas: normales etiquetados como outliers)")
print(f"Falsos Negativos (FN): {fn} (Anomalías que el modelo NO detectó)")

# Métricas clave en detección de anomalías
recall = tp / (tp + fn) if (tp + fn) > 0 else 0
precision = tp / (tp + fp) if (tp + fp) > 0 else 0

print(f"\nRecall (Sensibilidad): {recall:.2%} (De todos los outliers reales, ¿cuántos atrapamos?)")
print(f"Precision: {precision:.2%} (De todos los que predijimos como outliers, ¿cuántos lo eran realmente?)")
```

## Modelos basados en densidad

### DBSCAN (Density-Based Spatial Clustering of Applications with Noise) 

https://scikit-learn.org/stable/modules/generated/sklearn.cluster.DBSCAN.html

es un algoritmo excelente para detección de anomalías porque no asume que los clusters tienen forma esférica (como K-Means) y, lo más importante, tiene un concepto nativo de "ruido". Cualquier punto que no pertenezca a una región densa es etiquetado automáticamente como anomalía. A diferencia del Z-score o Tukey que evalúan características unidimensionales o distancias globales a un único centroide, DBSCAN busca anomalías locales evaluando la densidad del vecindario alrededor de cada punto.

```python
from sklearn.neighbors import NearestNeighbors
# Definir k (que será nuestro min_samples en DBSCAN)
k = 5
# Calcular la distancia a los vecinos más cercanos
# Usamos n_neighbors=k. (Nota: el vecino 0 es el punto mismo con distancia 0)
vecinos = NearestNeighbors(n_neighbors=k)
vecinos.fit(X_Escalado)
distancias, indices = vecinos.kneighbors(X_Escalado)

# Extraer y ordenar las k-distancias
# Tomamos la última columna, que corresponde a la distancia con el k-ésimo vecino
distancias_k = distancias[:, -1]
# Ordenamos las distancias de menor a mayor para formar la curva
distancias_k_ordenadas = np.sort(distancias_k)

# 5. Graficar el Scree Plot (Gráfico de K-distancias)
plt.figure(figsize=(10, 6))
plt.plot(distancias_k_ordenadas, color='blue', linewidth=2)

# Configuraciones visuales
plt.title(f'Gráfico de K-distancias para k={k} (Método del Codo)', fontsize=14)
plt.xlabel('Puntos ordenados por distancia', fontsize=12)
plt.ylabel(f'Distancia al {k}-ésimo vecino más cercano (eps)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.7)

punto_codo = 5
plt.axhline(y=punto_codo, color='red', linestyle='--', 
            label=f'Posible codo detectado (eps ≈ {punto_codo})')
plt.legend()

plt.tight_layout()
plt.show()
```

```python
from sklearn.cluster import DBSCAN

# Configurar y entrenar DBSCAN
# eps: Radio del vecindario. Si es muy grande, todo es un cluster. Si es muy chico, todo es ruido.
# min_samples: Cantidad mínima de puntos dentro del radio 'eps' para ser considerado un "core point" (región densa).
modelo_dbscan = DBSCAN(eps=5, min_samples=5)
# Entrenar el modelo y obtener las etiquetas
etiquetas = modelo_dbscan.fit_predict(X_Escalado)
# DBSCAN asigna la etiqueta -1 a los puntos que considera "Ruido" (Outliers)
# Las etiquetas >= 0 corresponden a los diferentes clusters encontrados.
n_clusters = len(set(etiquetas)) - (1 if -1 in etiquetas else 0)
n_ruido = list(etiquetas).count(-1)
porcentaje_anomalias = (n_ruido / len(X)) * 100

print("--- Resultados de DBSCAN ---")
print(f"Clusters encontrados: {n_clusters}")
print(f"Puntos anómalos (Ruido, etiqueta -1): {n_ruido}")
print(f"Porcentaje de anomalías: {porcentaje_anomalias:.2f}%")
```

```python
y_pred_DBSCAN = (etiquetas == -1).astype(int)
cm = confusion_matrix(y_true, y_pred_DBSCAN)
tn, fp, fn, tp = cm.ravel()

print("--- Resumen de la Clasificación ---")
print(f"Verdaderos Positivos (TP): {tp} (Anomalías reales detectadas)")
print(f"Verdaderos Negativos (TN): {tn} (Registros normales clasificados correctamente)")
print(f"Falsos Positivos (FP): {fp} (Falsas alarmas: normales etiquetados como outliers)")
print(f"Falsos Negativos (FN): {fn} (Anomalías que el modelo NO detectó)")

# Métricas clave en detección de anomalías
recall = tp / (tp + fn) if (tp + fn) > 0 else 0
precision = tp / (tp + fp) if (tp + fp) > 0 else 0

print(f"\nRecall (Sensibilidad): {recall:.2%} (De todos los outliers reales, ¿cuántos atrapamos?)")
print(f"Precision: {precision:.2%} (De todos los que predijimos como outliers, ¿cuántos lo eran realmente?)")
```

### LOF (Local Outlier Factor)

https://scikit-learn.org/stable/modules/generated/sklearn.neighbors.LocalOutlierFactor.html

LOF evalúa la desviación local de la densidad de un punto con respecto a la densidad de sus vecinos. La densidad local se evalúa por knn. 


```python
from sklearn.neighbors import LocalOutlierFactor
# Configurar y entrenar el modelo LOF
# n_neighbors: puntos vecinos para medir el desvío de la densidad.
# contamination: La proporción de outliers esperados en el dataset (ej. 0.05 para 5%). 
#                Si usas 'auto', Scikit-Learn calcula el umbral basado en el score.
modelo_lof = LocalOutlierFactor(n_neighbors=10, contamination=0.05)

# LOF devuelve 1 para inliers (normales) y -1 para outliers (anomalías)
etiquetas_lof = modelo_lof.fit_predict(X_Escalado)

# Extraer el "Outlier Score"
# negative_outlier_factor_ es negativo. Multiplicamos por -1 para que sea intuitivo: 
# Valores cercanos a 1 son normales (densidad similar a sus vecinos).
# Valores mucho mayores que 1 son anomalías (densidad menor que sus vecinos).
scores_anomalia = -modelo_lof.negative_outlier_factor_

# Resumen de resultados
n_anomalias = list(etiquetas_lof).count(-1)
print("--- Resultados de Local Outlier Factor (LOF) ---")
print(f"Total de registros: {len(X)}")
print(f"Anomalías detectadas: {n_anomalias} ({(n_anomalias/len(X))*100:.2f}%)")
```

```python
y_pred_LOF = (etiquetas_lof == -1).astype(int)
cm = confusion_matrix(y_true, y_pred_LOF)
tn, fp, fn, tp = cm.ravel()

print("--- Resumen de la Clasificación ---")
print(f"Verdaderos Positivos (TP): {tp} (Anomalías reales detectadas)")
print(f"Verdaderos Negativos (TN): {tn} (Registros normales clasificados correctamente)")
print(f"Falsos Positivos (FP): {fp} (Falsas alarmas: normales etiquetados como outliers)")
print(f"Falsos Negativos (FN): {fn} (Anomalías que el modelo NO detectó)")

# Métricas clave en detección de anomalías
recall = tp / (tp + fn) if (tp + fn) > 0 else 0
precision = tp / (tp + fp) if (tp + fp) > 0 else 0

print(f"\nRecall (Sensibilidad): {recall:.2%} (De todos los outliers reales, ¿cuántos atrapamos?)")
print(f"Precision: {precision:.2%} (De todos los que predijimos como outliers, ¿cuántos lo eran realmente?)")
```

## Modelos basados en subespacios
### Isolation Forest

https://scikit-learn.org/stable/modules/outlier_detection.html

El algoritmo construye un conjunto de árboles de decisión. En cada nodo, selecciona al azar una variable y un punto de corte entre su valor mínimo y máximo. Como los valores atípicos se encuentran en regiones poco densas del espacio de características, requieren menos cortes aleatorios para ser aislados (se encuentran más cerca de la raíz del árbol).

```python
from sklearn.ensemble import IsolationForest
# Configurar y entrenar Isolation Forest
# - n_estimators: Cantidad de árboles aleatorios (100 es el valor estándar recomendado).
# - contamination: Proporción de observaciones esperadas como anómalas (ej. 0.05 para 5%).
# - max_samples: Tamaño de la submuestra asignada a cada árbol ('auto' utiliza min(256, n_samples)).
iso_forest = IsolationForest(n_estimators=100, contamination=0.05, random_state=42)
iso_forest.fit(X_Escalado)

# 3. Predicciones y Scores de Aislamiento
# predict() asigna: 1 para registros normales, -1 para anomalías/outliers
etiquetas_if = iso_forest.predict(X_Escalado)

# decision_function() devuelve la puntuación de anomalía en el espacio centrado en 0.
# Puntuaciones negativas indican anomalías fuertemente aisladas.
score_decision = iso_forest.decision_function(X_Escalado)

# Invertimos el signo para obtener una métrica directa donde un valor mayor indica mayor nivel de anomalía
score_anomalia = -score_decision

y_pred_isoF = (etiquetas_if == -1).astype(int)

# 4. Métrica y resumen de ejecución
n_anomalias = y_pred_isoF.sum()
print("--- Resultados de Isolation Forest ---")
print(f"Total de registros analizados: {len(X)}")
print(f"Anomalías identificadas (-1): {n_anomalias} ({(n_anomalias/len(X))*100:.2f}%)")
print(f"Rango de Score de Anomalía: [{score_anomalia.min():.4f}, {score_anomalia.max():.4f}]")
```

```python
cm = confusion_matrix(y_true, y_pred_isoF)
tn, fp, fn, tp = cm.ravel()

print("--- Resumen de la Clasificación ---")
print(f"Verdaderos Positivos (TP): {tp} (Anomalías reales detectadas)")
print(f"Verdaderos Negativos (TN): {tn} (Registros normales clasificados correctamente)")
print(f"Falsos Positivos (FP): {fp} (Falsas alarmas: normales etiquetados como outliers)")
print(f"Falsos Negativos (FN): {fn} (Anomalías que el modelo NO detectó)")

# Métricas clave en detección de anomalías
recall = tp / (tp + fn) if (tp + fn) > 0 else 0
precision = tp / (tp + fp) if (tp + fp) > 0 else 0

print(f"\nRecall (Sensibilidad): {recall:.2%} (De todos los outliers reales, ¿cuántos atrapamos?)")
print(f"Precision: {precision:.2%} (De todos los que predijimos como outliers, ¿cuántos lo eran realmente?)")
```

# Propuesta: ECOD clínico (outliers por colas empíricas con dirección clínica)

## ¿Por qué Isolation Forest no alcanza en este problema?

Todos los métodos anteriores buscan puntos "raros" en **cualquier dirección** del espacio de 21 variables. Pero en este dataset lo que nos interesa detectar (NSP = sospechoso o patológico) **no es cualquier rareza**, sino desvíos en direcciones clínicamente conocidas:

- **Muchas variables no aportan señal de patología**: Min, Max, Mode, Mean, Median, Nmax, Nzeros, Tendency describen la forma del histograma de la FCF y son muy redundantes entre sí. Un registro puede ser raro en ellas siendo completamente normal, y eso "diluye" la señal (falsos positivos).
- **El desvío es unilateral**: un feto con ASTV del 80% (variabilidad anormal casi todo el tiempo) es preocupante; uno con ASTV del 15% no. Lo mismo para pocas aceleraciones (AC bajo) o muchas desaceleraciones (DL, DS, DP altos). IF, LOF, k-NN, Mahalanobis, etc. tratan ambas colas como igual de sospechosas.
- **Variables muy asimétricas y con muchos ceros** (DS, DP, FM, AC): las distancias euclídeas o las densidades quedan dominadas por la escala de unas pocas variables, aun después del `RobustScaler`.

## La idea

Usamos **ECOD** (*Empirical Cumulative distribution based Outlier Detection*, Li et al., 2022), un método no supervisado y sin hiperparámetros: para cada variable se calcula la probabilidad empírica de la cola, $p_j = P(X_j \geq x_{ij})$, y el score es $\sum_j -\log(p_j)$. Es decir, cuánto de "cola extrema" acumula cada registro.

Lo adaptamos al dominio: en lugar de mirar las dos colas de las 21 variables, miramos **solo la cola clínicamente preocupante** de las variables que las guías de interpretación de CTG (FIGO 2015) usan para clasificar un trazado:

| Variable | Cola de riesgo | Motivo clínico |
|---|---|---|
| ASTV | alta | % de tiempo con variabilidad de corto plazo anormal |
| ALTV | alta | % de tiempo con variabilidad de largo plazo anormal |
| MSTV | baja | variabilidad reducida → posible hipoxia |
| MLTV | baja | variabilidad reducida |
| AC | baja | ausencia de aceleraciones → trazado no reactivo |
| DL, DS, DP | alta | desaceleraciones (leves, severas, prolongadas) |
| LB | ambas | taquicardia (>160) o bradicardia (<110) basal |

**Importante**: las direcciones se definen *a priori* por conocimiento médico, **no** se ajustan mirando `NSP`. Las etiquetas solo se usan, igual que antes, para evaluar.

Ventajas adicionales: al trabajar con rangos (percentiles empíricos) no necesita escalado, es robusto a asimetrías y ceros, y el score es **interpretable**: podemos decir qué variable explica cada alarma.

```python
from scipy.stats import rankdata

# Direcciones de riesgo definidas a priori según criterios clínicos (FIGO), sin mirar las etiquetas
direcciones_clinicas = {
    'ASTV': 'alta',   # % de tiempo con variabilidad de corto plazo anormal
    'ALTV': 'alta',   # % de tiempo con variabilidad de largo plazo anormal
    'MSTV': 'baja',   # variabilidad de corto plazo media reducida
    'MLTV': 'baja',   # variabilidad de largo plazo media reducida
    'AC':   'baja',   # pocas aceleraciones -> trazado no reactivo
    'DL':   'alta',   # desaceleraciones leves
    'DS':   'alta',   # desaceleraciones severas
    'DP':   'alta',   # desaceleraciones prolongadas
    'LB':   'ambas',  # taquicardia o bradicardia basal
}

def cola_empirica(x, lado):
    # -log de la probabilidad empírica de observar un valor igual o más extremo en la cola indicada.
    # Usamos rangos con method='max' para que los empates (ej. muchos ceros) compartan la misma probabilidad.
    x = np.asarray(x, dtype=float)
    n = len(x)
    if lado == 'alta':
        return -np.log(rankdata(-x, method='max') / n)
    if lado == 'baja':
        return -np.log(rankdata(x, method='max') / n)
    # 'ambas': nos quedamos con la cola más extrema de las dos
    return np.maximum(cola_empirica(x, 'alta'), cola_empirica(x, 'baja'))

# Contribución de cada variable al score (útil para interpretar cada alarma)
contribuciones = pd.DataFrame({col: cola_empirica(X[col], lado) for col, lado in direcciones_clinicas.items()})

# Score ECOD clínico: suma de las "sorpresas" de cola de cada variable
score_ecod = contribuciones.sum(axis=1).values

# Mismo criterio de umbral que Isolation Forest (contamination=0.05) para comparar en igualdad de condiciones
umbral_ecod = np.percentile(score_ecod, 95)
y_pred_ecod = (score_ecod > umbral_ecod).astype(int)

print("--- Resultados de ECOD clínico ---")
print(f"Total de registros analizados: {len(X)}")
print(f"Umbral (percentil 95): {umbral_ecod:.2f}")
print(f"Anomalías identificadas: {y_pred_ecod.sum()} ({y_pred_ecod.mean()*100:.2f}%)")
```

```python
cm = confusion_matrix(y_true, y_pred_ecod)
tn, fp, fn, tp = cm.ravel()

print("--- Resumen de la Clasificación ---")
print(f"Verdaderos Positivos (TP): {tp} (Anomalías reales detectadas)")
print(f"Verdaderos Negativos (TN): {tn} (Registros normales clasificados correctamente)")
print(f"Falsos Positivos (FP): {fp} (Falsas alarmas: normales etiquetados como outliers)")
print(f"Falsos Negativos (FN): {fn} (Anomalías que el modelo NO detectó)")

# Métricas clave en detección de anomalías
recall = tp / (tp + fn) if (tp + fn) > 0 else 0
precision = tp / (tp + fp) if (tp + fp) > 0 else 0

print(f"\nRecall (Sensibilidad): {recall:.2%} (De todos los outliers reales, ¿cuántos atrapamos?)")
print(f"Precision: {precision:.2%} (De todos los que predijimos como outliers, ¿cuántos lo eran realmente?)")
```

### Comparación con Isolation Forest independiente del umbral

Precision y recall dependen del umbral elegido (5%). Para comparar los **scores** en sí usamos:

- **ROC-AUC**: probabilidad de que un registro anómalo reciba un score mayor que uno normal (0.5 = azar).
- **PR-AUC (Average Precision)**: más informativa cuando las clases están desbalanceadas. El valor de referencia de un score al azar es la proporción de anómalos (~22%).

Además evaluamos distintos umbrales. Notar que en este dataset **~22% de los registros son sospechosos o patológicos**: con `contamination=0.05` el recall nunca puede superar ~23%, así que el umbral también es una decisión que hay que justificar con el dominio.

```python
from sklearn.metrics import roc_auc_score, average_precision_score, precision_recall_curve

# score_anomalia viene de Isolation Forest (mayor = más anómalo)
scores_modelos = {'Isolation Forest': score_anomalia, 'ECOD clínico': score_ecod}

print(f"Proporción real de anómalos: {y_true.mean():.2%}\n")
print(f"{'Modelo':<18} | {'ROC-AUC':>7} | {'PR-AUC':>6} | {'Top 5%: P / R':>14} | {'Top 10%: P / R':>15} | {'Top 22%: P / R':>15}")
print("-" * 92)
for nombre, s in scores_modelos.items():
    fila = f"{nombre:<18} | {roc_auc_score(y_true, s):7.3f} | {average_precision_score(y_true, s):6.3f}"
    for pct in [5, 10, 22]:
        pred = s > np.percentile(s, 100 - pct)
        tp = (pred & (y_true == 1)).sum()
        fila += f" | {tp / pred.sum():6.0%} / {tp / y_true.sum():3.0%}  "
    print(fila)

# Curvas Precision-Recall
plt.figure(figsize=(8, 6))
for nombre, s in scores_modelos.items():
    prec, rec, _ = precision_recall_curve(y_true, s)
    plt.plot(rec, prec, linewidth=2, label=f'{nombre} (AP = {average_precision_score(y_true, s):.3f})')
plt.axhline(y_true.mean(), color='gray', linestyle='--', label='Azar (proporción de anómalos)')
plt.title('Curva Precision-Recall: Isolation Forest vs ECOD clínico', fontsize=14)
plt.xlabel('Recall', fontsize=12)
plt.ylabel('Precision', fontsize=12)
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

### Interpretabilidad: ¿por qué se marcó cada registro?

Una ventaja clave frente a Isolation Forest en un contexto médico: el score es una suma, así que podemos ver qué variable aporta más a cada alarma.

### Conclusiones

- En este problema la "anomalía" tiene un significado clínico. Incorporar ese conocimiento especifico de la clínica (qué variables importan y en qué dirección) mejora mucho más el resultado que cambiar el algoritmo genérico: probando otros métodos no supervisados genéricos (ECOD en las 21 variables, Mahalanobis robusta/MCD, reconstrucción por PCA, One-Class SVM) se obtienen ROC-AUC entre 0.5 y 0.77, en el mismo orden que Isolation Forest.
- ECOD clínico es **no supervisado**: no usa `NSP` para entrenar, no tiene hiperparámetros que ajustar y no requiere escalar los datos.
- Limitación: Depende de que las direcciones de riesgo sean correctas. En otro dominio, o si aparece un tipo de anomalía que no está en la lista, no lo detectaría. Una opción práctica es combinarlo con un detector genérico (como Islation Forest) para cubrir rarezas no previstas.
