# Atributos sin informacion (near zero variance)

```python
def check_nzv(dataframe, threshold=0.95):
    """
    Identifica columnas donde el valor más dominante excede el umbral dado.
    """
    nzv_cols = []
    for col in dataframe.columns:
        # Ignoramos nulos temporalmente para el cálculo de frecuencia de valores presentes
        if dataframe[col].dropna().empty:
            continue
            
        # value_counts con normalize=True devuelve la proporción (0 a 1)
        max_freq = dataframe[col].value_counts(normalize=True).iloc[0]
        
        if max_freq >= threshold:
            nzv_cols.append({'Columna': col, 'Max_Freq': max_freq, 'Top_Value': dataframe[col].mode()[0]})
            
    return pd.DataFrame(nzv_cols).sort_values(by='Max_Freq', ascending=False)

# Aplicamos la función a nuestro dataset
nzv_report = check_nzv(df, threshold=0.98)
print("Variables con Varianza Cercana a Cero (dominancia > 98%):")
display(nzv_report.head(10))
```

## 1. Tratamiento de Sótanos (Basements) - Ceros Estructurales

Primero identificamos las columnas relacionadas al sótano y las separamos en categóricas y numéricas.

```python
bsmt_cat_cols = ['BsmtQual', 'BsmtCond', 'BsmtExposure', 'BsmtFinType1', 'BsmtFinType2']
bsmt_num_cols = ['BsmtFinSF1', 'BsmtFinSF2', 'BsmtUnfSF', 'TotalBsmtSF', 'BsmtFullBath', 'BsmtHalfBath']

print("Nulos en variables categóricas de sótano:")
print(df[bsmt_cat_cols].isna().sum())
print("\nNulos en variables numéricas de sótano:")
print(df[bsmt_num_cols].isna().sum())

# Definimos "sin sótano" en base a la variable numérica más confiable: TotalBsmtSF
sin_sotano = df['TotalBsmtSF'] == 0
print(f"\nPropiedades sin sótano (TotalBsmtSF == 0): {sin_sotano.sum()}")
```

Verificamos que la cantidad de nulos en las variables categóricas coincida con la cantidad de propiedades sin sótano (`TotalBsmtSF == 0`). Esto nos da evidencia de que los NaN son **nulos estructurales** (la propiedad no tiene sótano) y no errores de carga.

```python
# Nulos categóricos vs. propiedades sin sótano, columna por columna
for col in bsmt_cat_cols:
    print(f"{col}: nulos={df[col].isna().sum()}, coincide con sin_sotano={(df[col].isna() == sin_sotano).all()}")

# Valores no nulos (deberían ser 0) en las columnas numéricas para las filas sin sótano
print("\nValores de las columnas numéricas en filas SIN sótano (deberían ser todos 0):")
print(df.loc[sin_sotano, bsmt_num_cols].sum())
```

`BsmtExposure` y `BsmtFinType2` tienen un nulo más que el resto (38 vs. 37). Eso significa que, para esas dos columnas, hay una propiedad con sótano (`TotalBsmtSF > 0`) cuyo valor categórico está vacío: un nulo **genuino**, no estructural. Lo aislamos:

```python
for col in bsmt_cat_cols:
    nulos_genuinos = df[col].isna() & ~sin_sotano
    if nulos_genuinos.any():
        print(f"{col}: nulo genuino (con sótano) en:")
        display(df.loc[nulos_genuinos, ['Id', 'TotalBsmtSF', col]])
```

Aplicamos el tratamiento:
* Categóricas en filas **sin sótano**: reemplazamos NaN por `'Sin_Sotano'`.
* Categóricas con nulo **genuino** (sótano existente, dato no registrado): imputamos con la moda de la columna entre las propiedades que sí tienen sótano, en vez de `'Sin_Sotano'` (que sería incorrecto, ya que esas propiedades sí tienen sótano).
* Numéricas: nos aseguramos que, para las filas identificadas como "sin sótano", el valor sea estrictamente `0` (y aplicamos `fillna(0)` por si quedara algún nulo suelto fuera de ese grupo).

Así garantizamos consistencia entre ambas familias de variables.

```python
# Tratamiento de las categóricas
for col in bsmt_cat_cols:
    df[col] = df[col].fillna('Sin_Sotano')

# Tratamiento de las numéricas: forzamos 0 en las filas sin sótano y limpiamos nulos remanentes
for col in bsmt_num_cols:
    df.loc[sin_sotano, col] = 0
    df[col] = df[col].fillna(0)

# Verificación final de consistencia
print("Nulos remanentes en variables de sótano:")
print(df[bsmt_cat_cols + bsmt_num_cols].isna().sum())

consistente = (df.loc[df['BsmtQual'] == 'Sin_Sotano', bsmt_num_cols] == 0).all().all()
print(f"\n¿Las filas 'Sin_Sotano' tienen todas sus variables numéricas en 0? {consistente}")
```

## 2. Auditoría de Inconsistencias

Construimos una función genérica que, dado un par (columna categórica "sin estructura", columna numérica), detecta filas donde la categórica indica ausencia de la estructura pero la numérica indica lo contrario (valor > 0). La aplicamos a `Pool` (y de paso a `Garage`, para mostrar que generaliza).

```python
def audit_inconsistencias(dataframe, cat_col, num_col, etiquetas_sin_estructura=()):
    """
    Detecta filas donde cat_col indica 'sin estructura' (NaN o alguna etiqueta dada)
    pero num_col es estrictamente mayor a 0.
    """
    sin_estructura = dataframe[cat_col].isna() | dataframe[cat_col].isin(etiquetas_sin_estructura)
    anomalia = sin_estructura & (dataframe[num_col] > 0)
    return dataframe.loc[anomalia, ['Id', cat_col, num_col]]

anomalias_pool = audit_inconsistencias(df, 'PoolQC', 'PoolArea')
print(f"Anomalías Pool (PoolQC nulo/sin piscina pero PoolArea > 0): {len(anomalias_pool)}")
display(anomalias_pool)

anomalias_garage = audit_inconsistencias(df, 'GarageType', 'GarageArea')
print(f"\nAnomalías Garage (GarageType nulo pero GarageArea > 0): {len(anomalias_garage)}")
display(anomalias_garage)
```

**Interpretación:** si `PoolArea > 0`, la propiedad *sí* tiene piscina; el NaN en `PoolQC` no es un nulo estructural en esas filas puntuales, sino un dato faltante real (no se registró la calidad). Tratarlo como `'Sin_Piscina'` sería incorrecto, porque distorsionaría la variable (diríamos que no hay piscina cuando sí la hay).

**Corrección propuesta:** para esas filas puntuales, imputamos `PoolQC` con la moda de `PoolQC` entre las propiedades que sí tienen piscina (en vez de con `'Sin_Piscina'`). Al resto de los nulos (donde `PoolArea == 0`, verdadero nulo estructural) los completamos con `'Sin_Piscina'`.

```python
# Moda de calidad entre las propiedades que efectivamente tienen piscina
moda_poolqc_con_piscina = df.loc[df['PoolArea'] > 0, 'PoolQC'].mode()[0]
print(f"Moda de PoolQC entre propiedades con piscina: {moda_poolqc_con_piscina}")

# Corrección de las filas anómalas (PoolArea > 0 y PoolQC nulo)
mask_anomalia = df['PoolQC'].isna() & (df['PoolArea'] > 0)
df.loc[mask_anomalia, 'PoolQC'] = moda_poolqc_con_piscina

# El resto de los nulos sí son nulos estructurales (no hay piscina)
df['PoolQC'] = df['PoolQC'].fillna('Sin_Piscina')

# Verificación: ya no deberían quedar anomalías
anomalias_pool_post = audit_inconsistencias(df, 'PoolQC', 'PoolArea', etiquetas_sin_estructura=['Sin_Piscina'])
print(f"Anomalías Pool tras la corrección: {len(anomalias_pool_post)}")
```

## 3. Métricas de Dispersión para NZV Numéricas

Para variables continuas, el "valor dominante" de `check_nzv` no es tan informativo como para categóricas. Usamos el **coeficiente de variación** ($CV = \sigma / \mu$): un CV muy bajo indica que casi todos los valores son parecidos entre sí (poca dispersión relativa a la media), es decir, poco poder discriminante.

Ignoramos los ceros estructurales (p.ej. `PoolArea = 0` para "sin piscina") porque inflarían artificialmente la cantidad de valores "iguales" y no reflejan variabilidad real de la estructura cuando esta existe.

```python
def check_nzv_numeric(dataframe, cv_threshold=0.05, excluir=('Id', 'SalePrice')):
    """
    Identifica columnas numéricas continuas con Near Zero Variance usando el
    coeficiente de variación (CV = std/mean), calculado ignorando los ceros
    estructurales (valores 0, que suelen representar "ausencia de estructura").
    """
    resultados = []
    numeric_cols = dataframe.select_dtypes(include=[np.number]).columns.difference(excluir)

    for col in numeric_cols:
        serie = dataframe[col].dropna()
        serie_no_cero = serie[serie != 0]

        if serie_no_cero.empty or serie_no_cero.mean() == 0:
            continue

        cv = serie_no_cero.std() / serie_no_cero.mean()
        resultados.append({
            'Columna': col,
            'CV': cv,
            'Media_no_cero': serie_no_cero.mean(),
            'Std_no_cero': serie_no_cero.std(),
            'N_no_cero': len(serie_no_cero)
        })

    resultado_df = pd.DataFrame(resultados).sort_values(by='CV')
    return resultado_df[resultado_df['CV'] < cv_threshold]

nzv_numericas = check_nzv_numeric(df, cv_threshold=0.05)
print("Variables numéricas con Varianza Cercana a Cero (CV < 0.05, ignorando ceros estructurales):")
display(nzv_numericas)
```
